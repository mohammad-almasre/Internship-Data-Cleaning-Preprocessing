# Data Preprocessing

## Missing Value 

#### Find Miss Values Or incorrect Values (country,director) 

```python
 string_cols = Netflix_df.select_dtypes(include=['object']).columns
Netflix_df[string_cols] = Netflix_df[string_cols].apply(lambda col: col.str.strip())
Netflix_df.replace('', np.nan, inplace=True)
```

```python
Netflix_df.isnull().sum()
```

```python
Netflix_df.info()
```

#### Replacing the incorrect data (not Given ) to (unknown)

```python
Netflix_df['director'] = Netflix_df['director'].replace('Not Given', 'Unknown')
Netflix_df['country'] = Netflix_df['country'].replace('Not Given', 'Unknown')
```

```python
print(Netflix_df['country'].value_counts().head(5))
```

## Duplicated Row

```python
print("Num of Dublicated Row:", Netflix_df.duplicated(subset=['title']).sum())
print("Num of Row:", len(Netflix_df))
```

```python
duplicates_df = Netflix_df[Netflix_df.duplicated(subset=['title'], keep=False)].sort_values(by='title')
duplicates_df
```

```python
Netflix_df = Netflix_df.drop_duplicates(subset=['title'], keep='first')
```

```python
print("Num of Dublicated Row:", Netflix_df.duplicated(subset=['title']).sum())
print("Num of Row:", len(Netflix_df))
```

## (Data Type Conversion)date_added

### convert Datatype from (object) to (datetime64) , To simplify the time analysis later.  

```python
Netflix_df['date_added'].head()
```

```python
Netflix_df['date_added'] = pd.to_datetime(Netflix_df['date_added'], errors='coerce')
```

```python
Netflix_df['date_added'].head()
```

## Extract Numarical Data from duration ( Moive , TV Show)

```python
Netflix_df['duration_min'] = Netflix_df[Netflix_df['type'] == 'Movie']['duration'].str.extract(r'(\d+)').astype(float)
Netflix_df['seasons_count'] = Netflix_df[Netflix_df['type'] == 'TV Show']['duration'].str.extract(r'(\d+)').astype(float)
```

```python
Netflix_df['duration_min'].head()
```

```python
Netflix_df['seasons_count'].head()
```

```python
Netflix_df['duration_min'] = Netflix_df['duration_min'].fillna(0).astype(int)
Netflix_df['seasons_count'] = Netflix_df['seasons_count'].fillna(0).astype(int)
```

```python
Netflix_df['duration_min'].head()
```

##### Delete 'duration' feature to avoid duplication

```python
Netflix_df.drop(columns=['duration'], inplace=True)
```

```python
Netflix_df.head()
```

## Extract (Year , Month ) from date_added to simplify analysis later

```python
Netflix_df['year_added'] = Netflix_df['date_added'].dt.year
Netflix_df['month_added'] = Netflix_df['date_added'].dt.month_name()
```

```python
Netflix_df['month_added'].head()
```
